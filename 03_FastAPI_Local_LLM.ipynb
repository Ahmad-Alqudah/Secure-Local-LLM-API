# 03 — FastAPI + Local LLM

## Objective
Build a REST API that communicates with a locally served LLM.

## Architecture

Client → FastAPI → Ollama → Qwen

## Components

### FastAPI
يستقبل الطلبات ويتحقق من البيانات ويعيد الاستجابات.

Handles API requests, validation, and responses.

### Pydantic
يتحقق من شكل البيانات وأنواعها.

Validates request data.

### Ollama
يشغّل نموذج Qwen محليًا.

Serves the local LLM for inference.

## API Endpoints

- GET /health
- POST /chat

## Error Handling

- Invalid input: 422
- Model server unavailable: 503
- Model response timeout: 504
- Unexpected model server failure: 502

## Key Principle

The backend controls the application interface,
while Ollama performs model inference.
